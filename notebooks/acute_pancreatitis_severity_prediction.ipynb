## The Problem

Acute pancreatitis is a sudden inflammation of the pancreas ranging from mild discomfort to life-threatening organ failure. **Early identification of severe cases is critical** — delayed treatment significantly increases mortality risk.

Current clinical scoring systems (e.g., Ranson, APACHE II) require 48-hour observation windows. This project demonstrates that a machine learning model trained on **routine admission blood tests** can predict severity at the point of first contact, enabling faster triage decisions without waiting.

## Dataset at a Glance

| Metric | Value |
|---|---|
| Real patients | **1,206** |
| Clinical features | **9** |
| Data source | Hospital clinical records |
| Target variable | Severity (Severe / Mild) |

**Biomarkers used:** CRP · Glucose · Creatinine · Calcium · WBC · Amylase · LDH · Albumin · Total Bilirubin

## Methodology

The project implements a complete end-to-end data pipeline:

**Raw Excel → ETL (Python) → MySQL → Random Forest → Power BI**

| Stage | Tool | Purpose |
|---|---|---|
| Extract & Clean | Python / pandas | Remove anomalies, standardize columns, filter clinically suspicious records (amylase < 50 U/L) |
| Storage | MySQL | Persist structured patient data for querying and BI connectivity |
| Prediction | Random Forest | Classify severity using 9 biomarkers; SMOTE applied to handle class imbalance |
| Visualization | Power BI | Interactive dashboard with real-time filters by sex and prediction type |

## Model Performance

### AUC-ROC: 0.92

> A score of 1.0 is a perfect classifier. **0.92 means the model correctly ranks a severe case above a mild case 92% of the time** — reaching diagnostic-grade accuracy on par with established clinical scoring systems.

| Class | Precision | Recall |
|---|---|---|
| Severe | 0.91 | 0.93 |
| Mild | 0.93 | 0.91 |

The model is well-balanced across both classes — it does not sacrifice sensitivity for specificity.

```python
import matplotlib.pyplot as plt

features = ['Calcium', 'Glucose', 'CRP', 'Bilirubin', 'LDH',
            'Albumin', 'WBC', 'Total Bilirubin', 'Creatinine']
importances = [0.22, 0.15, 0.13, 0.11, 0.10, 0.09, 0.08, 0.07, 0.05]

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.barh(features[::-1], importances[::-1], color='steelblue', height=0.6)

for bar, val in zip(bars, importances[::-1]):
    ax.text(bar.get_width() + 0.003, bar.get_y() + bar.get_height() / 2,
            f'{val:.0%}', va='center', fontsize=10, color='#333333')

ax.set_xlabel('Relative Importance', fontsize=11)
ax.set_title('Feature Importance for Severity Prediction', fontsize=13, fontweight='bold', pad=15)
ax.set_xlim(0, 0.27)
ax.spines[['top', 'right']].set_visible(False)
ax.tick_params(axis='y', labelsize=10)
plt.tight_layout()
plt.show()
```

## What the Model Found

The three most predictive biomarkers align with established clinical evidence:

- **Calcium (22%)** — Hypocalcemia is a direct marker of pancreatic necrosis and fat saponification. Its top ranking confirms it as the strongest early-warning signal for severe disease.
- **Glucose (15%)** — Stress hyperglycemia reflects systemic metabolic failure, a hallmark of severe acute pancreatitis.
- **CRP (13%)** — C-reactive protein is already used in Glasgow scoring; the model independently identifies it as a key discriminator.

The remaining biomarkers (LDH, albumin, bilirubin, WBC) collectively contribute 45%, confirming that **no single test is sufficient** — the model's strength lies in combining all nine.

## Conclusions
